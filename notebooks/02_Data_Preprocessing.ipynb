```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv(r"C:\Users\ADC\Desktop\ICU_MLOps_Project\data\raw\ICU.csv")

print("Dataset Loaded Successfully")
print(df.shape)

df.head()
```

```python
df = df.drop(columns=["Unnamed: 0", "ID"])

print(df.shape)

df.head()
```

```python
X = df.drop("Survive", axis=1)

y = df["Survive"]

print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
print("Feature Columns:")
print(X.columns.tolist())

print("\nTarget Column:")
print(y.name)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("="*50)
print("Training Data")
print("="*50)

print(X_train.shape)
print(y_train.shape)

print("\n")

print("="*50)
print("Testing Data")
print("="*50)

print(X_test.shape)
print(y_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(
X_train_scaled,
columns=X.columns,
index=X_train.index
)

X_test_scaled = pd.DataFrame(
X_test_scaled,
columns=X.columns,
index=X_test.index
)
```

```python
X_train_scaled.head()
```

```python
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)

print("Scaler Saved Successfully")
```

```python
train_processed = X_train_scaled.copy()

train_processed["Survive"] = y_train.values

train_processed.to_csv(
    "../data/processed/train_processed.csv",
    index=False
)
```

```python
test_processed = X_test_scaled.copy()

test_processed["Survive"] = y_test.values

test_processed.to_csv(
    "../data/processed/test_processed.csv",
    index=False
)
```

```python
print("="*60)
print("PREPROCESSING COMPLETED SUCCESSFULLY")
print("="*60)

print("Original Dataset :", df.shape)

print("Training Data :", X_train_scaled.shape)

print("Testing Data :", X_test_scaled.shape)

print("Scaler Saved : Yes")

print("Processed Data Saved : Yes")
```
